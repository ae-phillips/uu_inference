### Question 7

```python
# Question 7 set up

import pandas as pd
import numpy as np

df = pd.read_csv("./data/ames_prices.csv")

# create indicator
df["central_air"] = (df["Central.Air"] == "Y").astype(int)

df.head()
```

```python
# 7.a

display(df['central_air'].value_counts())

p_hat = df["central_air"].mean()
print(f"Proportion of homes with central air: {p_hat:.2f}")
```

```python
# 7.b

rng = np.random.default_rng(seed=100)

R = 2000
replications = []

for r in range(R):
    df_r = df.sample(
        frac=1.0,
        replace=True,
        random_state=rng
    )

    p_r = df_r["central_air"].mean()
    replications.append(p_r)

replications = np.array(replications)

print("Random number seed: 100", "\nBootstrap replications: 2000")
```

```python
# 7.c

bootstrap_se = np.std(replications, ddof=1)
print(f"Bootstrap standard error: {bootstrap_se:.4f}")
```

```python
# 7.d

ci90 = (
    p_hat - 1.645 * bootstrap_se,
    p_hat + 1.645 * bootstrap_se
)

ci95 = (
    p_hat - 1.96 * bootstrap_se,
    p_hat + 1.96 * bootstrap_se
)

ci99 = (
    p_hat - 2.576 * bootstrap_se,
    p_hat + 2.576 * bootstrap_se
)

print(f"90% CI: [{ci90[0]:.3f}, {ci90[1]:.3f}]")
print(f"95% CI: [{ci95[0]:.3f}, {ci95[1]:.3f}]")
print(f"99% CI: [{ci99[0]:.3f}, {ci99[1]:.3f}]")
```

```python

print(f"90% CI: [{ci90[0]:.3f}, {ci90[1]:.3f}] -- Contains 0.90? {ci90[0] <= 0.90 <= ci90[1]}")
print(f"95% CI: [{ci95[0]:.3f}, {ci95[1]:.3f}] -- Contains 0.90? {ci95[0] <= 0.90 <= ci95[1]}")
print(f"99% CI: [{ci99[0]:.3f}, {ci99[1]:.3f}] -- Contains 0.90? {ci99[0] <= 0.90 <= ci99[1]}")
```

**7.e interpretation:** We reject the null hypothesis (H0: p = 0.90, or the true population proportion of central air homes is 0.90). We reject the null because 0.90 does not appear in any of the three confidence intervals (it's below), so the data provide evidence that the true population proportion lies above 0.9.

### Question 8

```python
# question 8 set up: pearson correlation
area = np.arcsinh(df["area"])
price = np.arcsinh(df["price"])

rho_hat = area.corr(price)

print(f"Original correlation estimate: {rho_hat:.3f}")
```

```python
# 8.a

rng = np.random.default_rng(seed=100)

R = 2000
replications = []

for r in range(R):
    df_r = df.sample(
        frac=1.0,
        replace=True,
        random_state=rng
    )

    area_r = np.arcsinh(df_r["area"])
    price_r = np.arcsinh(df_r["price"])

    rho_r = area_r.corr(price_r)
    replications.append(rho_r)

replications = np.array(replications)
print("Random number seed: 100\nBootstrap replications: 2000")
```

```python
# 8.b

bootstrap_se = np.std(replications, ddof=1)

print(f"Bootstrap standard error: {bootstrap_se:.4f}")
```

```python
# 8.c

ci90 = (
    rho_hat - 1.645 * bootstrap_se,
    rho_hat + 1.645 * bootstrap_se
)

ci95 = (
    rho_hat - 1.96 * bootstrap_se,
    rho_hat + 1.96 * bootstrap_se
)

ci99 = (
    rho_hat - 2.576 * bootstrap_se,
    rho_hat + 2.576 * bootstrap_se
)

print(f"90% CI: [{ci90[0]:.3f}, {ci90[1]:.3f}]")
print(f"95% CI: [{ci95[0]:.3f}, {ci95[1]:.3f}]")
print(f"99% CI: [{ci99[0]:.3f}, {ci99[1]:.3f}]")
```

```python
# 8.d

print(f"90% CI: [{ci90[0]:.3f}, {ci90[1]:.3f}] -- Contains 0? {ci90[0] <= 0 <= ci90[1]}")
print(f"95% CI: [{ci95[0]:.3f}, {ci95[1]:.3f}] -- Contains 0? {ci95[0] <= 0 <= ci95[1]}")
print(f"99% CI: [{ci99[0]:.3f}, {ci99[1]:.3f}] -- Contains 0? {ci99[0] <= 0 <= ci99[1]}")
```

**8.d interpretation:** We reject the null hypothesis (the true population correlation is 0). We reject the null because 0 does not appear in any of the three confidence intervals (it's below), so the data provide evidence that the true population correlation is higher, suggesting a positive linear relationship between area and price.

**8.e:** Area and price must remain paired during resampling because correlation relies on pairing. You want to make sure you're always using the area and price from the same row. Otherwise, you could be comparing the area of one house to the price of a different house.

### Question 9

```python
# Question 9 set up
df = pd.read_csv("./data/metabric.csv")

survival = df["Overall Survival (Months)"].dropna()

mu_hat = survival.mean()

print(f"Original mean survival/follow-up time: {mu_hat:.3f} months")
```

```python
# 9.a

rng = np.random.default_rng(seed=100)

R = 2000
replications = []

for r in range(R):
    df_r = df.sample(
        frac=1.0,
        replace=True,
        random_state=rng
    )

    mu_r = df_r["Overall Survival (Months)"].mean()
    replications.append(mu_r)

replications = np.array(replications)

print("Random number seed: 100\nBootstrap replications: 2000")
```

```python
# 9.b

bootstrap_se = np.std(replications, ddof=1)

print(f"Bootstrap standard error: {bootstrap_se:.3f}")
```

```python
# 9.c

ci90 = (
    mu_hat - 1.645 * bootstrap_se,
    mu_hat + 1.645 * bootstrap_se
)

ci95 = (
    mu_hat - 1.96 * bootstrap_se,
    mu_hat + 1.96 * bootstrap_se
)

ci99 = (
    mu_hat - 2.576 * bootstrap_se,
    mu_hat + 2.576 * bootstrap_se
)

print(f"90% CI: [{ci90[0]:.3f}, {ci90[1]:.3f}]")
print(f"95% CI: [{ci95[0]:.3f}, {ci95[1]:.3f}]")
print(f"99% CI: [{ci99[0]:.3f}, {ci99[1]:.3f}]")
```

```python
# 9.d

print(f"90% CI: [{ci90[0]:.3f}, {ci90[1]:.3f}] -- Contains 120? {ci90[0] <= 120 <= ci90[1]}")
print(f"95% CI: [{ci95[0]:.3f}, {ci95[1]:.3f}] -- Contains 120? {ci95[0] <= 120 <= ci95[1]}")
print(f"99% CI: [{ci99[0]:.3f}, {ci99[1]:.3f}] -- Contains 120? {ci99[0] <= 120 <= ci99[1]}")
```

**9.d interpretation:** At all three CIs, we reject the null hypothesis (the true population mean is 120 months). We reject the null because 120 does not appear in any of the three confidence intervals (it's below), so the data provide evidence that the true population mean is above 120 months.

**9.e:**   

The estimated mean recorded time to death or last follow-up is approximately 129 months, or about 10.8 years.  

* The 90% confidence interval is approximately 125.658 to 132.531 months, or 10.47 to 11.04 years  

* The 95% confidence interval is approximately 125.000 to 133.189 months, or 10.42 to 11.10 years  

* The 99% confidence interval is approximately 123.713 to 134.476 months, or 10.31 to 11.21 years

These results do not estimate the unrestricted mean lifetime because some of the data refers to last follow-up, not death. This means the final value we have for some patients is not death, so we know their eventual death times will be to the right of that final appointment, but we can't estimate when.

```python
# 10.b

bootstrap_se = np.std(replications, ddof=1)

print(f"Bootstrap standard error: {bootstrap_se:.3f}")
```

```python
# 10.c

ci90 = (
    mu_hat - 1.645 * bootstrap_se,
    mu_hat + 1.645 * bootstrap_se
)

ci95 = (
    mu_hat - 1.96 * bootstrap_se,
    mu_hat + 1.96 * bootstrap_se
)

ci99 = (
    mu_hat - 2.576 * bootstrap_se,
    mu_hat + 2.576 * bootstrap_se
)

print(f"90% CI: [{ci90[0]:.3f}, {ci90[1]:.3f}]")
print(f"95% CI: [{ci95[0]:.3f}, {ci95[1]:.3f}]")
print(f"99% CI: [{ci99[0]:.3f}, {ci99[1]:.3f}]")
```
